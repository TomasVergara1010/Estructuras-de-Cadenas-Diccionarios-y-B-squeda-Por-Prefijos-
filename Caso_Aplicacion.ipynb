```python
import random
#CLASE PARA LOS NODOS DEL TRIE --> Esto es la base del trie
class TrieNode:
    def __init__(self):
        self.children = {}          # Array a las siguientes letras
        self.is_end_of_word = False # Marca si termina el nombre de la fruta


#CLASE PARA EL ÁRBOL DE PREFIJOS (TRIE)
class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self, word: str):
        """Inserta una palabra letra por letra en el árbol."""
        temp = self.root
        for char in word.lower():
            if char not in temp.children:
                temp.children[char] = TrieNode()
            temp = temp.children[char]
        temp.is_end_of_word = True

    def _dfs(self, temp: TrieNode, prefix: str, resultados: list):
    #Método auxiliar para buscar lo que sigue despues del prefijo, ejm: "man" --> "zana", "go","darina"
        if temp.is_end_of_word:
            resultados.append(prefix)
        for char, child in temp.children.items():
            self._dfs(child, prefix + char, resultados)

    def starts_with(self, prefix: str) -> list:
    #Devuelve todas las frutas que comiencen con el prefijo dado
        temp = self.root
        for char in prefix.lower():
            if char not in temp.children:
                return [] # Si la letra no existe, no hay sugerencias
            temp = temp.children[char]

        resultados = []
        self._dfs(temp, prefix.lower(), resultados)
        return resultados



if __name__ == "__main__":
    #Esto de las frutas serian las cadenas
    catalogo_frutas = [
        "manzana", "mango", "mandarina", "maracuyá", "mora", "melón", "melocotón",
        "pera", "piña", "papaya",
        "banano", "baya", "breva",
        "fresa", "frambuesa",
        "uva", "uva verde", "Uva Isabella",
        "sandía", "ciruela", "cereza", "kiwi", "limón", "limon mandarino", "naranja"
    ]

    #Creamos el Trie y metemos las frutas
    trie = Trie()
    for fruta in catalogo_frutas:
        trie.insert(fruta)

    #Creamos un Diccionario con sus precios
    precios_frutas = {
        "manzana": 2500, "mango": 3000, "mandarina": 2000, "maracuyá": 3500,
        "mora": 2200, "melón": 4500, "melocotón": 5000, "pera": 2800,
        "piña": 4000, "papaya": 3500, "banano": 1200,
        "baya": 6000, "breva": 3800, "fresa": 4200, "frambuesa": 7000,
        "uva": 5000, "uva verde": 5200, "uva isabella": 4800,
        "sandía": 6000, "ciruela": 3000, "cereza": 9000, "kiwi": 4500,
        "limón": 1500, "limon mandarino": 1800, "naranja": 2000
    }


    # SIMULACIÓN EN LA BARRA DE BÚSQUEDA
    prefijo_escrito = "m"  #-->Pal ejm: man, ma, uv, li,
    print(f"1. El cajero escribe el prefijo: '{prefijo_escrito}'")

    #Usamos el Trie para autocompletar
    sugerencias = trie.starts_with(prefijo_escrito)
    print(f"-> Sugerencias encontradas en el Trie ({len(sugerencias)}): {sugerencias}")

    #Si el cajero selecciona la primera sugerencia, consultamos el Diccionario
    if sugerencias:
        fruta_sorpresa = random.choice(sugerencias)
        print(f"El sistema elige una al azar: '{fruta_sorpresa.title()}'")
        precio = precios_frutas.get(fruta_sorpresa)
        print(f"-> Precio en el Diccionario: ${precio:,} COP")
```
