```python
import kagglehub
import os
import pandas as pd

path = kagglehub.dataset_download("karkavelrajaj/amazon-sales-dataset")

print("Los archivos se descargaron en la ruta:", path)
print(os.listdir(path))

archivo_csv = os.path.join(path, "amazon.csv")
df = pd.read_csv(archivo_csv)

print(df.head())

nombres_productos = df['product_name'].dropna().tolist()
print(f"Total de productos cargados: {len(nombres_productos)}")
```

```python
class TrieNode:
    def __init__(self):
        self._children = [None] * 256
        self._isEndOfWord = False


class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self, word: str):
        temp = self.root
        for char in word:
            indice = ord(char)
            if indice >= 256:
                continue

            if temp._children[indice] is None:
                temp._children[indice] = TrieNode()
            temp = temp._children[indice]
        temp._isEndOfWord = True

    def search(self, word: str):
        actual = self.root
        for char in word:
            indice = ord(char)
            if indice >= 256 or actual._children[indice] is None:
                return False
            actual = actual._children[indice]
        return actual._isEndOfWord

    def _dfs(self, temp: TrieNode, prefix: list, resultados: list):
        if temp._isEndOfWord:
            resultados.append("".join(prefix))

        for i in range(256):
            if temp._children[i] is not None:
                prefix.append(chr(i))
                self._dfs(temp._children[i], prefix, resultados)


    def startsWith(self, prefix: str):
        actual = self.root
        for char in prefix:
            indice = ord(char)
            if indice >= 256 or actual._children[indice] is None:
                return []
            actual = actual._children[indice]

        resultados = []
        self._dfs(actual, list(prefix), resultados)
        return resultados
```

```python
import random

#print("Ejemplos de productos en la base:", list(nombres_productos[:10]))
mi_trie = Trie()

for producto in nombres_productos:
    limpio = str(producto).strip().lower()
    mi_trie.insert(limpio)

print("--Bienvenido al Buscador Mas Duro De Amazon--")
prefijo_usuario = input("Escribe las primeras letras del producto: ")

sugerencias = mi_trie.startsWith(prefijo_usuario.lower())

if sugerencias:
    print(f"\nHay {len(sugerencias)} coincidencias.")
    producto_elegido = random.choice(sugerencias)

    print(f"Papi, se te recomienda este producto:")
    print(f"{producto_elegido}")

    detalles = df[df['product_name'] == producto_elegido]
    if not detalles.empty:
        precio = detalles.iloc[0].get('discounted_price', 'N/A')
        rating = detalles.iloc[0].get('rating', 'N/A')
        print(f"Precio: {precio}")
        print(f"Calificación: {rating}")
else:
    print("Nonas, no hay nada de eso (Por el momento)")
```
